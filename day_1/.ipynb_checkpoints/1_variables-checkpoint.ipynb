## Naming Conventions

1. Cannot Use Reserved Fields like (True, def)
2. Cannot Use Special characters ($,@,*)
3. Cannot Start variables by Numbers it should be _ or characters

```python
sample = "The Himalayas are one of the youngest mountain ranges on the planet."
# 1. ‘The Himalayas’ using slice operator
print(sample[:13])
# 2. Print "mountain range" using negative index
print(sample[-30:-14])
# 3. “The Himalayas on the planet” using slice as well as f-string
text1 = sample[:13]
text2 = sample[-14:]
print(f"{text1} {text2}")
```
