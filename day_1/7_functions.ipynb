## Functions

```python
def total(expenses):
    total_sum = 0
    for exp in expenses:
        total_sum += exp

    return total_sum

my_list = [20,23,45,9,7]

print(total(my_list))
```

## *args : Positional arguments

```python
def my_sample(*args):
    print(args)

# position arguments
my_sample(1,'hello',False)
```

## **kwargs : Keyword Arguments

```python
def my_sample(**kwargs):
    print(kwargs)

# position arguments
my_sample(a=1,b='hello',c=False)
```

```python
def my_sample(*args,**kwargs):
    print(args)
    print(kwargs)

my_sample('lalith','kumar',a=1,b='hello',c=False)
```
