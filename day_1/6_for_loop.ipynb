```python
expenses = [1200,1300,1500, 1700]

total_expense = 0
for expense in expenses:
    total_expense += expense

print(total_expense)
```

## Range

```python
total_expense = 0
for i in range(len(expenses)):
    expense = expenses[i]
    total_expense += expense
    print(f" Month of {i+1} : {expense}");

print(total_expense)
```

# Enumerate

```python
total_expense = 0
for i,expense in enumerate(expenses):
    total_expense += expense
    print(f" Month of {i+1} : {expense}");

print(total_expense)
```

## Break

```python
monthly_sales = [42, 38, 33, 38, 40, 45]

thresold = 35

for sales_amount in monthly_sales:
    if sales_amount < thresold:
        print(f"Sales amount {sales_amount} is less than the thresold")
        break
    else:
        print(f"sales amount {sales_amount} is greater than the thresold")
```

## Zip

```python
months = [ "jan","feb","mar","apr","may","jun"]

for sales_amount,month in zip(monthly_sales,months):
    if sales_amount < thresold:
        print(f"Sales amount {sales_amount} is less than the thresold in {month} month")
        break
    else:
        print(f"sales amount {sales_amount} is greater than the thresold in {month} month")
```
