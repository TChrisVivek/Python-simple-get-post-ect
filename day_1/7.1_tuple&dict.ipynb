## Dictionary

```python
sample_dict = {
    "lalith" : "+919035355000"
}

print(type(sample_dict))

sample_dict.get("lalith_kumar","-")
```

```python
sample_dic = {
    "1" : { "name" : "lalith", "phone" : " +919035355000" },
    "2" : { "name" : "kumar", "phone" : " +919035355001" },
    "3" : { "name" : "raja", "phone" : " +919035355002" }
}

for key,values in sample_dic.items():
    print(key , values.get('name') , values.get('phone'))

# Access Key
sample_dic['1']
# sample_dic['4'] # it will throw error exceptions
sample_dic.get('4') # will not throw exceptions
sample_dic.get('4','Not Found') # Get Default Value
```
